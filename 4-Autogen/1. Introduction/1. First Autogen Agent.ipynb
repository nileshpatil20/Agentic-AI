# First Autogen Agent


```python
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv('GOOGLE_API_KEY')
```

# Connecting to the Model

```python
# Create an instance of the OpenAIChatCompletionClient with the desired model and API key
model_client = OpenAIChatCompletionClient(model="gemini-1.5-flash-8b",api_key=api_key)

agent_1 = AssistantAgent(name='my_assistant',model_client=model_client)
result = await agent_1.run(task='what is capital of Maharashtra?')
print(result.messages[-1].content)
```

```python
result2 = await agent_1.run(task='What was my last question?')
print(result2.messages[-1].content)
```
